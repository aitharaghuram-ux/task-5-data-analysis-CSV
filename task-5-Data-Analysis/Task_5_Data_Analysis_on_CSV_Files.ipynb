##Load Data

```python
order_details = pd.read_csv("Order Details.csv")
```

```python
category_profit = order_details.groupby("Category")["Profit"].sum()
```

```python
category_profit
```

```python
category_profit.idxmax()
```

```python
category_profit.plot(kind="bar",title="Profit by Category")
```

```python
category_quantity = order_details.groupby("Category")["Quantity"].sum()
```

```python
category_quantity
```

```python
category_quantity.plot(kind="bar",title="Quantity by Category")
```

```python
subcategory_sales = order_details.groupby("Sub-Category")["Amount"].sum()
```

```python
subcategory_sales
```

```python
subcategory_sales.idxmax()
```

```python
subcategory_sales.plot(kind="bar",title="Sales by Sub-Category")
```

```python
orders = pd.read_csv("List of Orders.csv")
```

```python
sales_data = pd.merge(orders,order_details, on="Order ID")
```

```python
sales_data.head()
```

```python
state_sales = sales_data.groupby("State")["Amount"].sum()
```

```python
state_sales
```

```python
state_sales.idxmax()
```

```python
state_sales.plot(kind="bar",title="Sales by State")
```
